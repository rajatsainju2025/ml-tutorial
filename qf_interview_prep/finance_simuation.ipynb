```python
# P. Coin Piles

n_H = 2
n_T = 980
n_total = n_H + n_T
pile_size = n_H

# form two piles with equal number of heads
def simulate_pile_formation(n_total, n_H, n_T, pile_size):
    # create a random array with 1000 coins with 20 H
    coins = ['H'] * n_H + ['T'] * n_T
    print(coins)
    np.random.shuffle(coins)

    # take n coins from the
    p1 = coins[:pile_size]
    p2 = coins[pile_size:]

    # turn over all the coins in p_1
    p1 = ['T' if i=='H' else 'H' for i in p1]

    p1_H = p1.count('H')
    p2_H = p2.count('H')
    
    return p1_H, p2_H

simulate_pile_formation(n_total, n_H, n_T, pile_size)
```
